```python
def main():
    spacecraft = {"name":"james web space telescope" }
    spacecraft.update({"distance":0.01,"orbit":"sun"}) ## not just one key at once , to add multiplete keys at once i could use Update(),just like this
    print(create_report(spacecraft))
def create_report(spacecraft):
    return f"""
    ============Report========
    
    Name : {spacecraft.get("name")}
    Distance: {spacecraft.get("distance","unknown")}
    Orbit : {spacecraft["orbit"]}

    ==========================
    """
main()
```

```python
distances = {
    "voyagar1" : 163,
    "voyagar2" : 136,
    "pioneer 10"  : 80,
    "New Horizones":58,
    "pioneer 11" : 44
}
def convert(au):
    return au * 149,597,871 
def main():
    for distance in distances.values():
        print(f"{distance} is {convert(distance)}m")
main()
```
